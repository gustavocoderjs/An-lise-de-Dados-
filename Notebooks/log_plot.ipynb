```python
import numpy as np
import matplotlib.pyplot as plt

# Dados: crescimento exponencial (y = 10 * 2^x)
horas = np.arange(0, 10)  # De 0 a 9 horas
populacao = 10 * (2 ** horas)  # População inicial = 10, dobra a cada hora

print("Horas:", horas)
print("População:", populacao)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Dados
horas = np.arange(0, 10)
populacao = 10 * (2 ** horas)

# Plot
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.plot(horas, populacao, 'bo-')
plt.title("Escala Linear")
plt.xlabel("Horas")
plt.ylabel("População")
plt.grid()

plt.subplot(1, 2, 2)
plt.semilogy(horas, populacao, 'ro-')
plt.title("Escala Semi-Log (Y log)")
plt.xlabel("Horas")
plt.ylabel("População (log)")
plt.grid()

plt.tight_layout()
plt.show()
```
