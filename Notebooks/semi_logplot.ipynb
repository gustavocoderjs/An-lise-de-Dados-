```python
import numpy as np
import matplotlib.pyplot as plt

# Criando dados que seguem uma lei de potência: y = 2x³
x = np.linspace(1, 100, 100)
y = 2 * x**3

# Adicionando um pouco de ruído para parecer dados reais
y = y * np.random.normal(1, 0.1, len(x))

# Criando a figura
plt.figure(figsize=(12, 5))

# Gráfico normal (linear-linear)
plt.subplot(1, 2, 1)
plt.plot(x, y, 'bo')
plt.title('Escala Linear-Linear')
plt.xlabel('x')
plt.ylabel('y')

# Gráfico log-log
plt.subplot(1, 2, 2)
plt.loglog(x, y, 'bo')
plt.title('Escala Log-Log')
plt.xlabel('log(x)')
plt.ylabel('log(y)')
plt.grid(True, which="both", ls="--")

plt.tight_layout()
plt.show()
```
