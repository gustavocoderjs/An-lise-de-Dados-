```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
np.random.seed(42)
heights = np.random.normal(170, 10, 50)  # Mean 170cm, SD 10cm
weights = heights * 0.6 + np.random.normal(0, 5, 50)  # Roughly correlated with height

# Create scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(heights, weights, color='blue', alpha=0.7)
plt.title('Height vs. Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.grid(True)
plt.show()
```

```python
# Sample data for three groups
group_a_x = np.random.normal(5, 1, 30)
group_a_y = group_a_x * 2 + np.random.normal(0, 1, 30)

group_b_x = np.random.normal(8, 1, 30)
group_b_y = group_b_x * 1.5 + np.random.normal(0, 1, 30)

group_c_x = np.random.normal(3, 1, 30)
group_c_y = group_c_x * 0.5 + np.random.normal(0, 1, 30)

# Create scatter plot with multiple groups
plt.figure(figsize=(8, 6))
plt.scatter(group_a_x, group_a_y, color='red', label='Group A', alpha=0.6)
plt.scatter(group_b_x, group_b_y, color='green', label='Group B', alpha=0.6)
plt.scatter(group_c_x, group_c_y, color='blue', label='Group C', alpha=0.6)

plt.title('Scatter Plot with Multiple Groups')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from scipy import stats

# Generate some data with a linear relationship
x = np.arange(0, 10, 0.5)
y = 2 * x + 1 + np.random.normal(0, 1, len(x))

# Calculate regression line
slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
line = slope * x + intercept

plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='purple', label='Data points')
plt.plot(x, line, color='red', label=f'Regression line (y={slope:.2f}x+{intercept:.2f})')
plt.title('Scatter Plot with Regression Line')
plt.xlabel('Independent Variable')
plt.ylabel('Dependent Variable')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import UnivariateSpline, BSpline, make_interp_spline

# Gerar dados não-lineares com ruído
np.random.seed(42)
x = np.linspace(0, 10, 100)
y = np.sin(x) + np.random.normal(0, 0.2, 100)

# Ajustar diferentes tipos de splines
spline_suave = UnivariateSpline(x, y, s=5)  # Spline de suavização
spline_interp = make_interp_spline(x, y, k=3)  # Spline interpolador
x_fine = np.linspace(0, 10, 500)  # Grid mais fino para plotagem

# Plotar os resultados
plt.figure(figsize=(12, 6))
plt.scatter(x, y, color='gray', alpha=0.5, label='Dados originais')
plt.plot(x_fine, spline_suave(x_fine), 'r-', label='Spline de suavização (s=5)')
plt.plot(x_fine, spline_interp(x_fine), 'b--', label='Spline interpolador (k=3)')
plt.title('Comparação de diferentes tipos de Splines')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import make_smoothing_spline

# Gerar dados com padrão não-linear mais complexo
np.random.seed(42)
x = np.linspace(0, 10, 100)
y = np.sin(x*0.5) + 0.5*np.cos(x*2) + np.random.normal(0, 0.2, 100)

# Criar versões suaves com diferentes parâmetros
x_smooth = np.linspace(x.min(), x.max(), 300)

# Suavização 1 (mais suave)
spl1 = make_smoothing_spline(x, y, lam=0.1)
y_smooth1 = spl1(x_smooth)

# Suavização 2 (menos suave)
spl2 = make_smoothing_spline(x, y, lam=0.01)
y_smooth2 = spl2(x_smooth)

# Plotar os resultados
plt.figure(figsize=(12, 6))
plt.scatter(x, y, color='gray', alpha=0.5, label='Dados originais')
plt.plot(x_smooth, y_smooth1, 'r-', label='Suavização forte (λ=0.1)')
plt.plot(x_smooth, y_smooth2, 'b--', label='Suavização fraca (λ=0.01)')
plt.title('Alternativa usando Splines de Suavização do SciPy')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```
