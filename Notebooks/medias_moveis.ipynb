```python
def running_average(data, window_size):
    averages = []
    for i in range(len(data) - window_size + 1):
        window = data[i:i + window_size]
        avg = sum(window) / window_size
        averages.append(avg)
    return averages

# 🔢 Dados de exemplo: temperatura ao longo dos dias
temperaturas = [22, 23, 21, 20, 24, 25, 26]

# 🪟 Janela de 3 dias
media_movel = running_average(temperaturas, 3)

print("Médias móveis:", media_movel)
```

```python
import pandas as pd

# 🔢 Temperaturas ao longo dos dias
dados = {
    'dia': pd.date_range(start='2025-07-01', periods=7, freq='D'),
    'temperatura': [22, 23, 21, 20, 24, 25, 26]
}

df = pd.DataFrame(dados)

# 🪟 Calculando média móvel com janela de 3 dias
df['media_movel_3'] = df['temperatura'].rolling(window=3).mean()

print(df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Exemplo de dados
dados = {
    'data': pd.date_range(start='2025-07-01', periods=15, freq='D'),
    'vendas': [100, 120, 130, 90, 95, 140, 150, 160, 155, 145, 170, 175, 165, 180, 185]
}

df = pd.DataFrame(dados)
df.set_index('data', inplace=True)

# Suavização Exponencial
df['suav_exponencial'] = df['vendas'].ewm(alpha=0.3, adjust=False).mean()

# Plotando
plt.figure(figsize=(12, 6))
plt.plot(df.index, df['vendas'], label='Vendas Diárias', color='lightgray')
plt.plot(df.index, df['suav_exponencial'], label='Suavização Exponencial', color='blue')
plt.title('Suavização Exponencial das Vendas')
plt.xlabel('Data')
plt.ylabel('Quantidade')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
