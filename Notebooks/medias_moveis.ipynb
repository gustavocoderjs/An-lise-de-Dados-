```python
def running_average(data, window_size):
    averages = []
    for i in range(len(data) - window_size + 1):
        window = data[i:i + window_size]
        avg = sum(window) / window_size
        averages.append(avg)
    return averages

# 🔢 Dados de exemplo: temperatura ao longo dos dias
temperaturas = [22, 23, 21, 20, 24, 25, 26]

# 🪟 Janela de 3 dias
media_movel = running_average(temperaturas, 3)

print("Médias móveis:", media_movel)
```

```python
import pandas as pd

# 🔢 Temperaturas ao longo dos dias
dados = {
    'dia': pd.date_range(start='2025-07-01', periods=7, freq='D'),
    'temperatura': [22, 23, 21, 20, 24, 25, 26]
}

df = pd.DataFrame(dados)

# 🪟 Calculando média móvel com janela de 3 dias
df['media_movel_3'] = df['temperatura'].rolling(window=3).mean()

print(df)
```
