```python
import scipy.stats as stats
import numpy as np

# Gerando uma amostra aleatória de alturas (em metros)
np.random.seed(42)
amostra = np.random.normal(loc=1.72, scale=0.05, size=30)  # Média 1.72, Desvio padrão 0.05

# Definição da hipótese
media_teorica = 1.70

# Teste t
t_stat, p_valor = stats.ttest_1samp(amostra, media_teorica)

# Exibindo os resultados
print(f"Estatística do teste: {t_stat:.4f}")
print(f"p-valor: {p_valor:.4f}")

# Verificando significância com α = 0.05
alpha = 0.05
if p_valor < alpha:
    print(" Rejeitamos a hipótese nula: A média é significativamente diferente de 1.70m.")
else:
    print(" Não rejeitamos a hipótese nula: Não há evidências suficientes para afirmar que a média é diferente de 1.70m.")
```
