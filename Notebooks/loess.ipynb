```python
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

# Gerar dados de exemplo
np.random.seed(42)
x = np.linspace(0, 10, 100)
y = np.sin(x) + np.random.normal(0, 0.2, 100)

# Aplicar suavização LOESS
lowess = sm.nonparametric.lowess(y, x, frac=0.3)  # frac controla o grau de suavização

# Extrair os pontos suavizados
x_smooth = lowess[:, 0]
y_smooth = lowess[:, 1]

# Criar a plotagem
plt.figure(figsize=(10, 6))

# Plotar os dados originais
plt.scatter(x, y, color='blue', alpha=0.5, label='Dados Originais')

# Plotar a linha LOESS
plt.plot(x_smooth, y_smooth, color='red', linewidth=2, label='Suavização LOESS')

# Adicionar elementos ao gráfico
plt.title('Análise de Dados com Suavização LOESS', fontsize=14)
plt.xlabel('Variável X', fontsize=12)
plt.ylabel('Variável Y', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=12)
plt.tight_layout()

# Mostrar o gráfico
plt.show()
```
